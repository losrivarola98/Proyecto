Importaciones

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv('../output/df_final.csv')
```

Análisis univariado

`cantidad`

```python
cantidad_fecha = df.groupby('fecha').cantidad.sum()# cantidad de ventas por fecha
print("Media:",cantidad_fecha.mean())
print("Mediana:",cantidad_fecha.median())
print("Moda:",cantidad_fecha.mode().tolist())
print("Desvío estandar:",cantidad_fecha.std())
print("Coeficiente de variación:",cantidad_fecha.std()/cantidad_fecha.mean())
print("Mínimo:",cantidad_fecha.min())
print("Máximo:",cantidad_fecha.max())
print("Rango intercuartílico:",cantidad_fecha.quantile(0.75)-cantidad_fecha.quantile(0.25))
percentiles = cantidad_fecha.quantile([0.05,0.25,0.5,0.75,0.95])
print("Percentiles:")
print("5%: ",percentiles[0.05])
print("25%:",percentiles[0.25])
print("50%:",percentiles[0.5])
print("75%:",percentiles[0.75])
print("95%:",percentiles[0.95])

```

Se observa que no hay moda, ya que cada fecha tiene diferentes cantidades vendidas.

```python
plt.figure(figsize=(15, 7))
plt.plot(cantidad_fecha.index, cantidad_fecha.values, marker='o', linestyle='-')
plt.axhline(cantidad_fecha.mean(), color='red', linestyle='--', label=f'Media: {cantidad_fecha.mean():.2f}')
plt.axhline(cantidad_fecha.median(), color='green', linestyle=':', label=f'Mediana: {cantidad_fecha.median():.2f}')
plt.title('Cantidad de Ventas por Fecha')
plt.xlabel('02/01/2021 - 30/09/2021')
plt.ylabel('Cantidad')
plt.grid(True)
plt.tick_params(axis='x', labelbottom=False)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=cantidad_fecha)
plt.title('Diagrama de Caja de Cantidad de Ventas por Fecha')
plt.show()
```

`desc_secciones`

```python
pd.set_option('display.float_format', '{:.7f}'.format)
totales = df.groupby('desc_secciones').cantidad.sum().sort_values()
secciones = pd.DataFrame({
    'absoluta':totales,
    'relativa':totales / totales.sum(),
    'acumulada':(totales / totales.sum()).cumsum(),
})
secciones
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    secciones['absoluta'],
    labels=secciones.index,
    autopct='%1.1f%%',
    startangle=30
)

plt.title('Porcentaje de ventas por sección')
plt.axis('equal')
plt.show()
```

`desc_sub_secciones`

```python
totales = df.groupby('desc_sub_secciones').cantidad.sum().sort_values()
subsecciones = pd.DataFrame({
    'absoluta':totales,
    'relativa':totales / totales.sum(),
    'acumulada':(totales / totales.sum()).cumsum(),
})
subsecciones
```

```python
frecuencias = subsecciones['absoluta'].copy()
cantidad_otros = (subsecciones['relativa'] < 0.02).sum()
etiquetas = subsecciones.index.to_series().where(
    subsecciones['relativa'] >= 0.02,
    f'Otros ({cantidad_otros})'
 )
frecuencias_agrupadas = frecuencias.groupby(etiquetas.values).sum()

plt.figure(figsize=(8, 8))

plt.pie(
    frecuencias_agrupadas,
    labels=frecuencias_agrupadas.index,
    autopct='%1.1f%%',
)

plt.title('Porcentaje de ventas por Subsección')
plt.axis('equal')
plt.show()
```

Análisis bivariado

Variables numéricas (`cantidad`, `minima`, `media`, `maxima` y `precipitacion`)

```python
datos_correlacion = pd.concat(
    [cantidad_fecha, minima, media, maxima, precipitacion],
    axis=1
)

datos_correlacion.columns = [
    'cantidad', 'minima', 'media', 'maxima', 'precipitacion'
]

matriz_correlacion = datos_correlacion.corr(method='pearson')

plt.figure(figsize=(9, 7))
sns.heatmap(
    matriz_correlacion,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    square=True
)

plt.title('Mapa de calor de correlación de Pearson')
plt.tight_layout()
plt.show()
```

No se observan relaciones significativas entre cantidad y los demás, tampoco entre precipitacion y los demás. Evidentemente hay relaciones fuertes entre mínima, media y máxima, lo cual es esperable.

`secciones` vs. `temperatura media`

```python
categorias = pd.cut(df['media'], bins=[-float('inf'), 10, 20, 30, float('inf')], labels=['< 10°', '10° - 20°', '20° - 30°', '> 30°'])

secciones_temperatura = df.cantidad.groupby([categorias, df['desc_secciones']]).sum().unstack(fill_value=0).T
display(secciones_temperatura)
chi2, p_value, dof, expected = chi2_contingency(secciones_temperatura)
n = secciones_temperatura.sum().sum()
min_dim = min(secciones_temperatura.shape) - 1
cramers_v = np.sqrt(chi2 / (n * min_dim))

print(f"Chi-cuadrado: {chi2}")
print(f"p-value: {p_value}")
print(f"V de Cramér: {cramers_v}")
```

La asociación entre las secciones y la temperatura media es muy baja.

`secciones` vs `precipitacion`

```python
categorias = pd.cut(df['precipitacion'], bins=[-0.01, 0, 10, 25, 50, float('inf')], labels=['Sin lluvia', 'Ligera', 'Moderada', 'Fuerte', 'Muy fuerte'])

secciones_precipitacion = df.cantidad.groupby([categorias, df['desc_secciones']]).sum().unstack(fill_value=0).T
display(secciones_precipitacion)
chi2, p_value, dof, expected = chi2_contingency(secciones_precipitacion)
n = secciones_precipitacion.sum().sum()
min_dim = min(secciones_precipitacion.shape) - 1
cramers_v = np.sqrt(chi2 / (n * min_dim))

print(f"Chi-cuadrado: {chi2}")
print(f"p-value: {p_value}")
print(f"V de Cramér: {cramers_v}")
```

La asociación entre las secciones y las precipitaciones es muy baja.
